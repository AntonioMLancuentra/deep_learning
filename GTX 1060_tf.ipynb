```python
# This enables logging of where (which device) TensorFlow operations are placed — such as on the CPU or GPU — during execution.
import tensorflow as tf
tf.debugging.set_log_device_placement(True)
# How to turn it off: tf.debugging.set_log_device_placement(False)

with tf.device('/GPU:0'):
    a = tf.random.normal([1000, 1000])
    b = tf.random.normal([1000, 1000])
    c = tf.matmul(a, b)
    print("Matrix multiplication result:", c)
```

```python
import tensorflow as tf
import time
tf.keras.backend.clear_session()

def benchmark_matmul(device, num_iters=10):
    start = time.perf_counter()
    print(f"\nRunning on {device}...")
    with tf.device(device):
        # Try a more modest tensor size
        size = 11000  # Adjust based on what your GPU can handle
        a = tf.random.normal([size, size])
        b = tf.random.normal([size, size])

        @tf.function
        def run_matmul():
            return tf.matmul(a, b)
        
        for _ in range(num_iters):
            run_matmul()

    end = time.perf_counter()
    print(f"Time on {device}: {end - start:.4f} seconds")

gpus = tf.config.list_physical_devices('GPU')
if gpus:
    benchmark_matmul('/GPU:0')
#benchmark_matmul('/CPU:0')

# Running on /GPU:0...
# Time on /GPU:0: 1.0984 seconds

# Running on /CPU:0...
# Time on /CPU:0: 187.3298 seconds
```

```python
import torch

print("Is CUDA available? ", torch.cuda.is_available())

if torch.cuda.is_available():
    print("GPU Name:", torch.cuda.get_device_name(0))
    print("Device count:", torch.cuda.device_count())
    print("Current device:", torch.cuda.current_device())
    print("Torch CUDA version:", torch.version.cuda)
    print("Torch compiled with:", torch.__config__.show())
else:
    print("No GPU detected.")
```
